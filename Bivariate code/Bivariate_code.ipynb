```python
import pandas as pd
import seaborn as sns
```

```python
diabetes = pd.read_csv('diabetes.csv')
```

```python
diabetes.head()
```

```python
sns.scatterplot(x=diabetes['Glucose'],y=diabetes['BMI'],hue=diabetes['Outcome'],style=diabetes['Pregnancies'],size=diabetes['Age'])
```

```python
sns.barplot(x=diabetes['Pregnancies'],y=diabetes['Age'],hue=diabetes['Outcome'])
```

```python
sns.boxplot(x=diabetes['Outcome'],y=diabetes['Age'],hue=diabetes['Insulin'])
```

```python
sns.distplot(diabetes[diabetes['Outcome']==0]['Age'],hist=False)
sns.distplot(diabetes[diabetes['Outcome']==1]['Age'],hist=False)
```

```python
sns.heatmap(pd.crosstab(diabetes['Pregnancies'],diabetes['Outcome']))
```

```python
(diabetes.groupby('BMI').mean()['Outcome']*100)
```

```python
sns.clustermap(pd.crosstab(diabetes['Pregnancies'],diabetes['Outcome']))
```

```python
sns.pairplot(diabetes,hue='BMI')
```

```python
new = diabetes.groupby('Age').sum().reset_index()
```

```python
sns.lineplot(x=new['Age'],y=new['Pregnancies'])
```
